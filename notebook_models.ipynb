```python
%load_ext autoreload 
%autoreload 2

import numpy as np 
from matplotlib import pyplot as plt 
import models 

## general code for visualising models as they are defined in the module 'models' 
## may save to a current directory 

def visualise_model(model, Nx=100, Ny=100, save=True, contours=False):
    x_min, x_max = model.x_min, model.x_max 
    y_min, y_max = model.y_min, model.y_max

    ## compute on a discretised grid 
    x = np.linspace(x_min, x_max, Nx)
    y = np.linspace(y_min, y_max, Ny)
    z = model(x[:, None], y[None, :])

    ## plot 
    plt.figure() 
    pmesh = plt.pcolormesh(x, y, z)
    plt.colorbar(pmesh)
    plt.xlabel("$x$")
    plt.ylabel("$y$")
    plt.title(model.name)

    if save:
        save_name = model.name.replace(" ", "_").lower()
        plt.savefig(f"figures/models/{save_name}.pdf")
```
